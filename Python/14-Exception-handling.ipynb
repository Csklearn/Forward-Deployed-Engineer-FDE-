### Exception Handling

- try , except , else , finally, raise


```python
# Not handling exception
a = 10
b = 0
print(a/b)
```

```python
try:
    a = 10
    b = 0
    print(a/b)
except:
    print("zero division not allowed error")
```

```python
## using Exception
try:
    a = 10
    b = 0
    print(a/b)
except Exception as e:
    print(e)
```

```python
try:
    a = int(input("Enter a first number"))
    b = int(input("Enter a second number"))
    print(a/b)
except ZeroDivisionError:
    print("Zero Division Error")
except TypeError:
    print("Incorrect data type")
except Exception as e:
    print("Generic Error",e)
else:
    print("no exception occurs")
finally:
    print("Execution completed")
```

```python
# custom raise exception

age = int(input("Enter you age"))
try:
    if age < 18:
        raise ValueError("Not eligible to Vote")
    else:
        print("Eligible to vote")
# except Exception as e:
#     print("Exception occures")
finally:
    print("Excution completed")
```
