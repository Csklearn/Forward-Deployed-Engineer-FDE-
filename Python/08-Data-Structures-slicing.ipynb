# Slicing

Slicing means extracting the specific portion of content from string,list,tuple etx..

start - Starting index included
ends - end index excluded
Positive step - Start should be less than end  means left to right , Incremental slicing
Negative step - start should be less than end. right to left , decremental slicing
By default step is set to 1

```python
# Example of string slicing
# format variable[start:stop:step]

language = "PYTHON"
print(language[1:3])
print(language[1:5])
print(language[0:6:2])
```

```python
# Example for negative indexing
language = "PYTHON"
print(language[-1:-4:-1])
print(language[-1:-4:-2])
print(language[-1:-7:-1])
```

```python
# Example of incorrect start and stop values
language = "PYTHON"
print(language[3:1]) # here step is not provided, so it will take default value of 1, and since start is greater than stop, it will return empty string
print(language[-1:5:-1]) # here step is not provided, so it will take default value of 1, and since start is greater than stop, it will return empty string
```

```python
# Example without providing start and stop values
language = "PYTHON"
print(language[:]) # here start and stop values are not provided, so it will take default values of 0 and len(language) respectively, and it will return the whole string
```

```python
language = "PYTHON"
print(language[1:]) # here both start and stop values are provided, and it will return the string from index 1 to index 4 (exclusive)
print(language[:4]) # here both stop values is provided, and it will return the string from index 0 to index 4 (exclusive)
```
