# Operators 

- Arithmetic +,-,/,*,%,**
- Assignement +=,-+,*=.,%=
- Comparison >,<,>=,<=,!
- Logical and,or,not
- Bitwise |,&,^,~,>>,<<
- Identity is , is not

```python
# Arithmetic operators

a=10
b=20
print("Addition:",a+b)
print("Subtraction:",a-b)
print("Multiplication:",a*b)
print("Division:",a/b) ## return float value
print("Modulus:",a%b)  # return remainder
print("Exponentiation:",a**b) # return a raised to the power of b
print("Floor Division:",a//b) # return quotient without decimal value and near to negative infinity(Math.floor in JS)
print("Floor Division:",-1//2) # return quotient without decimal value and near to negative infinity(Math.floor in JS)
```

```python
a=10
b=20
a +=b  # a=a+b
print("Addition:",a)
a -=b  # a=a-b
print("Subtraction:",a)
a *=b  # a=a*b 
print("Multiplication:",a)
a /=b  # a=a/b
print("Division:",a) 
a %=b  # a=a%b
print("Modulus:",a)
a **=b  # a=a**b  
print("Exponentiation:",a)  
```

```python
# Comparison operators
a=10
b=20
print("Equal:",a==b)
print("Not Equal:",a!=b)
print("Greater than:",a>b)
print("Less than:",a<b)
print("Greater than or equal to:",a>=b)
print("Less than or equal to:",a<=b) 
```

```python
# Logical operators and or not
a=10
b=20
print("Logical AND:",a>5 and b>15)  # True and True => True
print("Logical OR:",a>5 or b<15)  # True or False => True
print("Logical NOT:",not(a>5 and b>15))  # not(True and True) => not(True) => False
```
