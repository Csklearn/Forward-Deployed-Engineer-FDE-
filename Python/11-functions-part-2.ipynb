# function with parameter and return value

- It will take input parameter
- return key word return value
- returned value from the function can be input to another function


```python
# example with function with arguments and return values

def set_numbers(a,b):
    return a+b
result = set_numbers(5, 10)
print(result)
```

```python
def calculate_bill(price,qty,discount):
    total_amount = price * qty
    discount_amount= total_amount * discount/100
    final_amount = total_amount - discount_amount
    return final_amount

total_bill= calculate_bill(100, 5, 10)
print(total_bill)    
```

```python
# return multiple values from a function. 
def calculate_bill(price,qty,discount):
    total_amount = price * qty
    discount_amount= total_amount * discount/100
    final_amount = total_amount - discount_amount
    return total_amount, discount_amount, final_amount

total, discount, final = calculate_bill(100, 5, 10) 
print(total, discount, final)
# return a tuple value
```

```python
# keyword arguments

def calculate_bill(price,qty,discount):
    total_amount = price * qty
    discount_amount= total_amount * discount/100
    final_amount = total_amount - discount_amount
    return final_amount

total_bill= calculate_bill(price=100,discount=10,qty=5) # by using keyword argument , we can skip order of the parameters. 
print(total_bill)    
```

```python
# global and local variables

marks=100 # global variable
def update_marks():
    marks=200 # local variable
    print(marks)

print(marks) # global variable
update_marks() # local variable
```

```python
marks=100 # global variable
def update_marks():
    global marks
    marks=200 # create marks as global variable and update the value of marks to 200
    print(marks)

print(marks) # global variable
update_marks() # local variable
```

```python
# keyword and positional arguments

def student_details(name,*marks,**address):
    print(name)
    print(marks)
    print(address)

student_details("John", 90, 80, 70, city="New York", country="USA")

## format def fn_name(positional_args, *args, **kwargs):
## *args is used to pass a variable number of positional arguments to a function. It allows you to pass any number of arguments, which are then collected into a tuple.
## **kwargs is used to pass a variable number of keyword arguments to a function. It allows you to pass any number of keyword arguments, which are then collected into a dictionary.
```
