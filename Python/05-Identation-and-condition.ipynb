# Indentataion and condition (if,else and elif)
#### Indentation always gives space for statement coming under conditions since its belongs to that condition

# if , else , elif

#### Semicolon always ends with every if,else,elif
#### Indentation is importan
#### else , elif cannot be standalone

```python
# if example
age =int(input("Enter your age:"))
if age>=18:
    print("You are eligible to vote.")
```

```python
# if and else example
age = int(input("Enter your age:"))
if age>=18:
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote.")
```

```python
# if else example and importance of missing elif
## Exceute all matches and print all the grades
marks = int(input("Enter your mark:"))
if marks>=90:
    print("You are Grade A")
if marks>= 70:
    print("You are Grade B")
if marks>=40:
    print("You are Grade C")
else:
    print("You are Fail")
```

```python
# if ,elif and else example
## Execute only the first match and print the grade
marks = int(input("Enter your mark:"))
if marks>=90:
    print("You are Grade A")
elif marks>= 70:
    print("You are Grade B")
elif marks>=40:
    print("You are Grade C")
else:
    print("You are Fail")
```

```python
# Nested condition

marks = int(input("Enter your mark:"))
lab = int(input("Enter your lab mark:"))
if marks>=80:
    if lab>=15:
        print("You are Grade A")
    else:
        print("You are Grade B")
else:
    print("You are Grade C")
```
