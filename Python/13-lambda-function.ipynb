# Lambda function

### Its anonymous function and it will take arguments and return the expression

### format lambda parameter : expression



```python
### Funcion using def keyword
def add_numbers(a,b):
    return a+b

result = add_numbers(5, 10)
print(result)  # Output: 15

#### Function using lambda keyword

add_numbers = lambda a,b: a+b
print(add_numbers(5, 10))  # Output: 15
```

```python
## Examples of lambda functions

result = lambda x: "even" if x % 2 == 0 else "odd"
print(result(6))  # Output: even
print(result(7))  # Output: odd
```

```python
result = lambda x: x**2
print(result(5))  # Output: 25
```

## eval 
### eval function parse the string input to int or float output

```python
# exmample for eval

num = eval(input("Enter your age"))
if num >= 18:
    print("You are eligible to vote")
else:
    print("You are not eligible to vote")
```

# warlus operator
### Its used to assign a expression output to a variable. so it reduce number of lines of code

```python
# Exampl for warlus operator

## Without warlue operator

items =[1,2,3,4]
count = len(items)
if count >= 3:
    print("Number of count is greater than or equal to 3")
else:
    print("Number of count is lesser than 3")
    

## Above exampl using warlus operator

list_items = [1,2,3,4]

if (count := len(list_items)) >=3:
        print("Number of count is greater than or equal to 3")
else:
    print("Number of count is lesser than 3")
    

    
```

```python
if (number := eval(input("Enter a number"))) > 10:
    print("Number is greater than 10")
else:
    print("Number is lesser than 10")
    
```

```python
is_count_greater_10=lambda x:x>10 
if is_count_greater_10(count := eval(input("Enter a number"))):
    print(count)
    print("count is greater than 10")
else:
    print("count is lesser than 10")
    
```

# Rule of lambda function compare with normal function

## lambda function
- DOesn't have functio name
- doesnt allow multiple statements
- doesnt allow mutiple assignment
- dont have return keywords sice expression returns



```python
calc= lambda x : (
    a=x+2
    b=x/2
    c=a+b
    return c
)

print(calc(4))
```

```python
# Example for *args , **kwargs

- Normal parameter can come before args
- Parameters comes after args are keyword only
- keyword only parameters can be swappes
- kwargs paramete must be last
- No parameter shouldnt come after kwargs
```
