```python
### Class withou encapsulation

class Account:
    def __init__(self,name,acctNo,balance=0):
        self.name=name
        self.acctNo=acctNo
        self.balance=balance
    
    def deposit(self,amount):
        self.balance += amount
        
    def withdraw(self,amount):
        self.balance -=amount
    
    def balance_check(self):
        print(f" balance -{self.acctNo} - {self.balance}")
        
acct_one = Account("MSD",123,1200)
acct_one.deposit(1000)
print(acct_one.balance_check())
acct_one.balance=0 ## because of accessible balnace set to o outised the class
print(acct_one.balance_check())
```

```python
### Class with encapsulation

class Account:
    def __init__(self,name,acctNo,balance=0):
        self.name=name
        self.acctNo=acctNo
        self.__balance=balance
    
    def deposit(self,amount):
        self.__balance += amount
        
    def withdraw(self,amount):
        self.__balance -=amount
    
    def balance_check(self):
        print(f" balance -{self.acctNo} - {self.__balance}")
        
acct_one = Account("MSD",123,1200)
acct_one.deposit(1000)
print(acct_one.balance_check())
acct_one.__balance=0 ## because of private its not accessible
print(acct_one.balance_check())
```

```python
## Inheritance - reusability
## Polymorphism - same name but corresponding child method is execueted
## super() is used to access parent class

class Payment:
    def __init__(self,payment_type,actNo,balance=0):
        self.payment_type=payment_type
        self.actNo=actNo
        self.balance=balance
    def pay(self,amount):
        self.amount=amount
        print(f"amount has been Paid {amount}")

class UPI(Payment):
    def __init__(self,payment_type,actNo,balance):
        super().__init__(payment_type,actNo,balance)
    def pay(self,amount):
        self.amount=amount
        print(f"amount has been paid through {self.payment_type} and amount is {self.amount}")

class Cash(Payment):
    def __init__(self,payment_type,actNo,balance):
        super().__init__(payment_type,actNo,balance)
    def pay(self,amount):
        self.amount=amount
        print(f"amount has been paid through {self.payment_type} and amount is {self.amount}")
        
upi_pay=UPI("upi",1234,8000)
upi_pay.pay(2000)

cash_pay=Cash("cash",1234,8000)
cash_pay.pay(2000)
    
```
