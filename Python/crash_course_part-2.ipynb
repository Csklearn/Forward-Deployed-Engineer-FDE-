```python
# the order of parameter

# position , default , multiple positions , multiple keywords
# default parameter created once and shared across instances

def add_item(items,cart=[]):
  cart.append(items)
  return cart
print(add_item("orange"))
print(add_item("banana"))
## here list get appen for each function call
def add_item_update(items,cart=None):
  if cart is None:
    cart=[]
  cart.append(items)
  return cart
print(add_item_update("orange"))
print(add_item_update("banana"))
```

```python
from typing import ParamSpecArgs
from abc import abstractmethod
# Encapsulation - Private data which cannot accessible outsid ethe class
class Student:
  def __init__(self,name,age,marks) -> None: # dunnter method - constructor self is the object which is passed
    self.name=name
    self.age=age
    self.marks=marks
    self.__id=1234 # private data
  def has_passed(self):
    return self.marks>40
  def get_id(self):
    return self.__id

student_one = Student("sachin",23,50)
print(student_one.name,student_one.age)
print("passed",student_one.has_passed())
print(student_one.get_id())
# print(student_one.__id) # access private not access

# Inheritance and Polymorphism

class Person:
  def __init__(self,name,age) -> None:
    self.name=name
    self.age=age
  def greet(self):
    print("Hello")

class Teacher(Person):
  def __init__(self,name,age,marks) -> None:
    super().__init__(name,age)
    self.marks=marks
  def greet(self):
    print(f"Teacher - {self.name}-{self.marks}")

teacher_one = Teacher("sachin",23,50)
teacher_one.greet()

# Abstarction

class Payments:
  @abstractmethod
  def Pay(self,amount):
    pass

class UPI(Payments):
  def Pay(self,amount):
    print(f"UPI - {amount}")

class Cash(Payments):
  def Pay(self,amount):
    print(f"Cash - {amount}")

upi = UPI()
upi.Pay(100)
cash = Cash()
cash.Pay(100)
```

```python
# dunnter method with str,repr,len,eq,Propert ,setter
class Product:
  def __init__(self,name,price):
    self.name=name
    self.price=price
  # def __str__(self) -> str:
  #   return f"Product - {self.name}-{self.price}"
  def __repr__(self) -> str:
    return f"Product({self.name})"
  def __len__(self):
    return len(self.name)
  def __eq__(self,other):
    return self.name==other.name
  @property
  def price(self):
    return round(self._price)
  @price.setter
  def price(self,value):
    if value<0:
      raise ValueError("price is invalid")
    self._price=value

product=Product("keyboard",1800)
product.price=1000
print(product)

# classmethod another way to create object
class Product:
  discount=10
  def __init__(self,name,price):
    self.name=name
    self.price=price
  @classmethod
  def from_csv(cls,lines):
    name,marks=lines.split(",")
    return cls(name,int(marks))
  @staticmethod
  def is_integer(num):
    return num%1==0
product=Product.from_csv("keyboard,1800")
print(product.name,product.price)
Product.is_integer(10)

```

```python
# Decorators Time , log and retry any function

# the function takes another function as an input and execute.

def wrapper(func):
  def say_hello():
    print("function start printing")
    func()
    print("function ends printing")
  return say_hello

@wrapper
def hello():
  print("Hello")
## hello = wrapper(hello)
print(hello())

## @wraps keep original function name
```
