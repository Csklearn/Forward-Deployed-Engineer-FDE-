# OOP

- It is a progamming approach to organise data and object
- It is a blue print to create a object
- It is used to avoid code reuse , readability

```python
### Example
## class is a keyword
## Student - class name and it should start with capital letter
## name , age are data
class Student:
    name="MSD"
    age= 28
    

studentOne_details=Student()
print(studentOne_details.name,studentOne_details.age)
```

```python
## Clas with data and function def
## __init__ dunder method which is used to set the value to data memeber
## __init__ is like constructor
## self , the object which we are creating. it can be in any name
## user defined methods should have self as first parameter

class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def printStudent(self):
        print(f"{self.name} - {self.age}")

student_one=Student("MSD",28)
student_one.printStudent()
```
