```python
# infinite loop
while True:
    print("This is an infinite loop.")

a=1
while a>=1:
    print("This is an infinite loop.")
    a+=1
```

```python
# chat bot to respond to user input upto 3 times
i=1
while i<=3:
    user_input = input("Enter your message (or type 'exit' to quit): ")
    if user_input.lower() == "exit":
        print("Chatbot: Goodbye!")
        break
    else:
        print("Chatbot: You said -", user_input)
    i += 1
if i > 3:
   print("Chatbot: You have reached the maximum number of messages. Goodbye!")    
```

```python
# break and continue in while loop 
# continue statement is used to skip the current iteration of the loop and move on to the next iteration. When the continue statement is encountered, the remaining code inside the loop for that iteration is skipped, and the loop proceeds to the next iteration.

a=1
while a<=10:
    if a%2==0:
        continue
    else:
        print(a," is odd number")
        a +=1
    a +=1

```
