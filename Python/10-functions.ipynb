```python
# no parameters and no return value
# def keyword def is used to define a function in Python. The function name is followed by parentheses, which may include parameters. The function body is indented and contains the code that will be executed when the function is called.
# colon is used to indicate the start of the function body. The return statement is used to return a value from the function, but it is optional. If no return statement is provided, the function will return None by default.

def print_hello():
    print("Hello, World!")
```

```python
# with parameters and return value
def get_age(age):
    if age >= 18:
        return "You are an adult."
    else:
        return "You are a minor."

print(get_age(20))
print(get_age(15))
```
