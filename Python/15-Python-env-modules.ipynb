# Python environments

- Common environment approach
  - requirements.txt contains all dependencies 
  - pip install -r requirements.txt or install dependencies without requirements.txt
  - read env variable from env file
  - run app by python filename.py

- Virtual environment approach
- issue with common approach if multiple app run on same vm
  Python version conflicts
  Environment conflicts
  OS level conflict issue
  patch level issue

- command
  python -m venev myenv
  Source naviaget to myenv/Scripts/activate.bat
  pip install -r requirments.txt
  echo venv
  deactivate
  activate

- UV environment approach
  python -m pip install uv
  uv init
  uv add openai or requiremnets.txt( uv pip install -r requirements.txt)
  python __init__.py
