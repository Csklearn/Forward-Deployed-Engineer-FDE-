```python
from pymongo import MongoClient

# Connect to a local instance
client = MongoClient("mongodb://localhost:27017/")

# Or connect to MongoDB Atlas (replace with your actual connection string)
# client = MongoClient("mongodb+srv://<username>:<password>@cluster.mongodb.net/?retryWrites=true&w=majority")

# Verify the connection
client.admin.command("ping")
print("Connected successfully!")
```

```python
# Access a database and a collection
db = client["movies"]
movies_list = db["imdb"]
```

```python
movies_list.insert_one({"name": "Three Idiots", "year": "2010",})
movies_list.insert_many([
    {"name": "Rabde", "year": "2006"},
    {"name": "MSD", "year": "2008"}
])

# READ: Find one document or loop through multiple matches
movie = movies_list.find_one({"name": "MSD"})
print("Found user:", movie)
```

```python
client.close()
```
