# Sqllite
- Sqllite is a relational light weight file based database
- Sqllite is available through built-in feature in python
- Doesnt require seperate installation

```python
import sqlite3
# Python module which helps to connect to sqlite db
```

```python
connection = sqlite3.connect("movies.db")
print(connection)
## create database and create it if not exist
```

```python
cur=connection.cursor()
## cursor object which helps to execute the query
```

```python
# create Table
cur.execute("""
CREATE TABLE MOVIES(
    id INTEGER PRIMARY KEY,
    name TEXT,
    genre TEXT
)
            """)
connection.commit()
## commit the changes to db
```

```python
cur.execute("""
INSERT INTO MOVIES (name,genre)
values ("Rabde","emotion") ,("Three idiots","comedy")
            """)
connection.commit()
```

```python
# retireve data from table
records = cur.execute(" SELECT * from MOVIES")
result = records.fetchall()
print(result)
```

```python
# Dynamic 

query = """
INSERT INTO MOVIES (name,genre)
values (?,?)
            """
name="MSD"
genre="Historic"


cur.execute(query,(name,genre))
connection.commit()
```

```python
records = cur.execute(" SELECT * from MOVIES")
result = records.fetchall()
print(result)
```

```python
cur.execute("CREATE TABLE moviebkp(title, year, score)")
connection.commit()
```

```python
## retieve tables
res = cur.execute("SELECT * FROM sqlite_master")
res.fetchall()
```

```python
cur.execute("""
    INSERT INTO moviebkp VALUES
        ('Monty Python and the Holy Grail', 1975, 8.2),
        ('And Now for Something Completely Different', 1971, 7.5)
""")
connection.commit()
data = [
    ("Monty Python Live at the Hollywood Bowl", 1982, 7.9),
    ("Monty Python's The Meaning of Life", 1983, 7.5),
    ("Monty Python's Life of Brian", 1979, 8.0),
]
cur.executemany("INSERT INTO moviebkp VALUES(?, ?, ?)", data)
connection.commit()  # Remember to commit the transaction after executing INSERT.
```

```python
for row in cur.execute("SELECT year, title FROM moviebkp ORDER BY year"):
    print(row)
```

```python
connection.close()
## close db connection
```
