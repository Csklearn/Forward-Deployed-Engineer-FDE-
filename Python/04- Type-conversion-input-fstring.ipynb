# Membership Operator , Type conversion , input and fstring

```python
# Membership Operator

student_marks=[30,50,60,70,80,90,100]
print("Is 50 in student_marks:",50 in student_marks)  # True 
print("Is 40 in student_marks:",40 in student_marks)  # False
print("Is 40 not in student_marks:",40 not in student_marks)  # True
```

```python
# type conversion
a="10"
print("type of a:",type(a).__name__)
b=int(a)
print("type of b:",type(b).__name__)
print("value of b:",b)
c=list(a)
print("type of c:",type(c).__name__)
print("value of c:",c)
d= float(a)
print("type of d:",type(d).__name__)
print("value of d:",d)
```

```python
# input and f string

username=input("Enter your username: ")
age=input("Enter your age: ")
print(f"Hello {username}, you are {age} years old. So you are {'' if int(age) >= 18 else 'not '}eligible to vote.")
```
