# Bitwise and Data type

- Bitwise |,&,^,~,>>,<<


```python
# bitwise
a=10
b=2
print("Bitwise AND:",a&b)  # 10 & 2 => 1010 & 0010 => 0010 => 2
print("Bitwise OR:",a|b)  # 10 | 2 => 1010 | 0010 => 1010 => 10
print("Bitwise XOR:",a^b)  # 10 ^ 2 => 1010 ^ 0010 => 1000 => 8
print("Bitwise NOT:",~a)  # ~10 => ~1010 => 0101 => 5
print("Bitwise Left Shift:",a<<1)  # 10 << 1 => 1010 << 1 => 10100 => 20 (add a 0 at the end)
print("Bitwise Right Shift:",a>>1)  # 10 >> 1 => 1010 >> 1 => 0101 => 5 (remove the last bit)
```
