```python
# for loope we can iterate over a sequence of numbers, strings, lists, tuples, dictionaries, and sets. The for loop is used to execute a block of code repeatedly for each item in the sequence. The syntax of the for loop is as follows:

marks=[90, 80, 70, 60, 50, 40, 30]
for mark in marks:
    print(mark)
```

```python
training_centre_name="Naresh IT Solutions"
for char in training_centre_name:
    print(char)
```

```python
marks=(90, 80, 70, 60, 50, 40, 30)
for mark in marks:
    print(mark)
```

```python
state_code ={"AP":"Andhra Pradesh", "TN":"Tamil Nadu", "KA":"Karnataka"}
for code in state_code:
    print(code, state_code[code])

for details in state_code.items():
    print(details)

for code in state_code.keys():
    print(code)
    
for value in state_code.values():
    print(value)
```

```python
for i in range(5):
    print(i)

for i in range(1, 10):
    print(i)

for i in range(1, 10, 2): # start from 1, end at 10, increment by 2, exlucding 10
    if i%2==0:
        print(i, "is even")
    else:
        print(i, "is odd")
```

```python
import random
match_found=False
for i in range(5):
    c=random.randint(1, 10)
    if c==i:
        match_found=True
        print("Match found, You won!")
        break
if not match_found:
    print("No match found. Try again!")
```
