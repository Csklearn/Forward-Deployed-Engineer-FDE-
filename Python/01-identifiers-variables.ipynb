# Identifier and Variables

- Use numbers , underscore and digits
- Don't start identifiers with digits
- All identifiers are vaiables but not all variables are identifiers (ex: function name)
- Don't use python keywords,builtins,hypen(-) or spaces in identifiers
- Case sensitive


```python
# Identifiers and Variables

score=10
print(score)

# use of _
server_name="localhost"
print(server_name)

server_error_400="Bad Request"
print(server_error_400)

```

```python
import keyword
keyword_list=keyword.kwlist
print(keyword_list)
print("if" in keyword_list)
print("len" in keyword_list)
print(keyword.iskeyword("if"))
print(keyword.iskeyword("len"))
```

```python
# use of keywords as identifiers and result in error
assert=10
print(assert)
```

```python
import builtins
print(dir(builtins))
print(dir(__builtins__))
print("if" in dir(builtins))
print("len" in dir(builtins))
```

```python
# get_list_length is an identifier and not a variable.
def get_list_length(my_list):
    return len(my_list)

print(get_list_length([1,2,3,4,5]))
```
