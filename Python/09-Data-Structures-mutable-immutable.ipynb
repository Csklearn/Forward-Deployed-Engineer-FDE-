# Immutable

- string
- tuple

# Mutable
- list , set, dictionary

```python
# Immutable

language = "PYTHON"
#language[0] = "p" # This will raise an error because strings are immutable in Python
print(language) # This will print "PYTHON" because the original string is unchanged
language = "python" # This is allowed because we are creating a new string and assigning it to the variable language
print(language) # This will print "python" because the variable language now refers to the new string
```

```python
# Immutable tuple
language_tuple = ("PYTHON", "JAVA", "C++")
#language_tuple[0] = "python" # This will raise an error because tuples are immutable in Python
print(language_tuple) # This will print the original tuple
language_tuple = ("python", "JAVA", "C++") # This is allowed because we are creating a new tuple and assigning it to the variable language_tuple
print(language_tuple) # This will print the new tuple
language_tuple[0] = "python" # This will raise an error because tuples are immutable in Python
print(language_tuple) # This will print "python" because we are accessing the first element of the new tuple
```

# tuple,list,set - allows indexing , slicing and sorting

```python
language = ["PYTHON", "JAVA", "C++"]
print("slicing list:", language[0:2]) # This will print the first two elements of the list
print("sorting list:", sorted(language)) # This will print the sorted list
print("reversing list:", language[::-1]) # This will print the reversed list
print("sort list in place:", language.sort()) # This will sort the list in place and return None
```

```python
# tuple example
language_tuple = ("PYTHON", "JAVA", "C++")
print("slicing tuple:", language_tuple[0:2]) # This will print the first two elements of the tuple
print("sorting tuple:", sorted(language_tuple)) # This will print the sorted tuple
print("reversing tuple:", language_tuple[::-1]) # This will print the reversed tuple
```

```python
# Mutable add , update , delete , remove ,extend,pop,insert examples
## tuple and string are immutable, so we cannot add, update, delete, remove, extend, pop or insert elements in them. But list is mutable, so we can perform these operations on list.

language = ["PYTHON", "JAVA", "C++"]
language.append("JavaScript") # Add an element to the end of the list
print("After append:", language)
language[0] = "python" # Update an element at a specific index
print("After update:", language)
language.remove("JAVA") # Remove a specific element from the list
print("After remove:", language)
language.pop() # Remove the last element from the list
print("After pop:", language)
language.insert(1, "HTML") # Insert an element at a specific index
print("After insert:", language)
language.extend(["CSS", "SQL"]) # Extend the list by appending elements from another list
print("After extend:", language)
language.append(["React", "Node.js"]) # Append a list as a single element
print("After append list:", language)
language.remove(["React", "Node.js"]) # Remove the list that was appended
print("After remove list:", language)
del language[0] # Delete an element at a specific index
print("After delete:", language)
language.insert(0, "C#") # Insert an element at the beginning of the list
print("After insert at beginning:", language)
language.pop(1) # Remove the element at index 1
print("After pop at index 1:", language)

```

```python
# set example and allow unique values only, so if we try to add duplicate values, it will not be added to the set.  
# indexing is not allowed in set, so we cannot access elements by index. But we can check if an element is in the set or not.    
language_set = {"PYTHON", "JAVA", "C++"}
print("Initial set:", language_set)
language_set.add("JavaScript") # Add an element to the set
print("After add:", language_set)
language_set.add("PYTHON") # Try to add a duplicate element
print("After trying to add duplicate:", language_set)
print("Check if 'JAVA' is in set:", "JAVA" in language_set) # Check if an element is in the set
language_set.remove("C++") # Remove an element from the set
print("After remove:", language_set)
language_set.pop() # Remove and return an arbitrary element from the set
print("After pop:", language_set)
```

```python
# dictionary example and allow unique keys only, so if we try to add duplicate keys, it will update the value of the existing key.
language_dict = {"Python": 1, "Java": 2, "C++": 3}
print("Initial dictionary:", language_dict)
language_dict["JavaScript"] = 4 # Add a new key-value pair to the dictionary
print("After adding new key-value pair:", language_dict)
language_dict["Python"] = 5 # Update the value of an existing key
print("After updating value of existing key:", language_dict)
language_dict.pop("Java") # Remove a key-value pair from the dictionary
print("After removing key-value pair:", language_dict)
```
